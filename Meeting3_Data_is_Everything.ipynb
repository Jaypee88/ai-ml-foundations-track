# 🌍 Section 1: Why Data is the Core of ML
#### *No model outperforms the quality of its data*

---

There is a widely repeated truth in the Machine Learning industry:

> *'Garbage in, garbage out.'*

No matter how sophisticated your algorithm is, a poorly understood, poorly cleaned,  
or poorly curated dataset will produce a model that is unreliable, biased, or outright wrong.  
The inverse is equally true — **a clean, well-understood dataset with a simple model  
will almost always outperform a complex model trained on bad data.**

### Where Practitioners Actually Spend Their Time

| Activity | % of Time (Industry Survey) |
|----------|-----------------------------|
| Collecting & loading data | 19% |
| Cleaning & preprocessing data | 26% |
| Exploratory data analysis | 18% |
| Feature engineering | 14% |
| Model training & tuning | 15% |
| Deployment & monitoring | 8% |

**Over 75% of ML work happens before a single model is trained.**  
This session is not a detour — it is the heart of the discipline.

### What We Mean by 'Data is Everything'

- 📦 **Volume** — Do you have enough samples to generalise?
- 🎯 **Relevance** — Do your features actually relate to what you're predicting?
- 🧹 **Quality** — Are there missing values, errors, duplicates, inconsistencies?
- ⚖️ **Balance** — Are your classes represented fairly, or is one massively overrepresented?
- 🕐 **Recency** — Is the data still representative of the world today?
- 🔏 **Ethics** — Is the data collected fairly, with consent, and free of harmful bias?

Every one of these questions will be answered during **Exploratory Data Analysis (EDA)** —  
which is what this entire session is about.

# 🗂️ Section 2: Types of Data & Dataset Anatomy
#### *Know your data before you touch it*

---

Before you load a single file, you need to understand the *type* of data you're working with.  
Different data types require different preprocessing pipelines, models, and evaluation strategies.

### By Structure

| Type | Description | Examples | Common ML Models |
|------|-------------|---------|------------------|
| **Tabular / Structured** | Rows and columns — like a spreadsheet | CSV, SQL tables | Linear models, Trees, XGBoost |
| **Text / NLP** | Free-form natural language | Reviews, emails, tweets | Transformers, LSTMs, TF-IDF |
| **Image** | Pixel grids (2D or 3D) | Photos, X-rays, satellite | CNNs, Vision Transformers |
| **Audio** | Time-series waveforms | Speech, music | RNNs, CNNs on spectrograms |
| **Time-Series** | Ordered sequences over time | Stock prices, sensor data | LSTMs, ARIMA, Prophet |
| **Graph** | Nodes & edges | Social networks, molecules | GNNs |

> 🔑 **This cohort focuses on tabular data** — the most common type in industry,  
> and the ideal foundation before moving to images and text in Meetings 8 & 9.

---

### By Variable Type (Critical for Preprocessing)

| Variable Type | Description | Examples | How to Handle |
|---------------|-------------|---------|---------------|
| **Continuous (Float)** | Any numeric value in a range | Age, income, temperature | Scale, normalise, bin |
| **Discrete (Int)** | Countable whole numbers | Number of children, visits | May treat as continuous or categorical |
| **Ordinal Categorical** | Categories with meaningful order | Rating (1–5), education level | Label encode or ordinal encode |
| **Nominal Categorical** | Categories with no inherent order | Gender, city, colour | One-hot encode |
| **Binary** | Exactly two values | Yes/No, 0/1, True/False | Usually fine as-is or map to 0/1 |
| **DateTime** | Timestamps | Purchase date, login time | Extract year, month, day, weekday, hour |

**Understanding variable types determines every preprocessing decision you make.**


````python
# ── Create our working dataset for this entire session ──────────────────────
# A realistic employee attrition dataset (will employees leave the company?)

np.random.seed(42)
n = 400

departments   = ['Engineering','Sales','HR','Finance','Marketing','Operations']
education_lvl = ['High School','Bachelor','Master','PhD']
job_levels    = ['Junior','Mid','Senior','Lead','Manager']

dept_arr  = np.random.choice(departments, n, p=[0.30,0.25,0.10,0.12,0.13,0.10])
edu_arr   = np.random.choice(education_lvl, n, p=[0.15,0.45,0.30,0.10])
level_arr = np.random.choice(job_levels, n, p=[0.25,0.30,0.25,0.12,0.08])

age           = np.random.randint(22, 60, n)
years_company = np.clip(np.random.randint(0, 20, n), 0, age - 22)
salary        = (30000 + age * 800 + np.random.normal(0, 8000, n)).clip(25000, 180000).round(-2)
satisfaction  = np.random.choice([1,2,3,4,5], n, p=[0.08,0.15,0.30,0.32,0.15])
perf_rating   = np.random.choice([1,2,3,4], n, p=[0.05,0.20,0.55,0.20])
overtime      = np.random.choice([0,1], n, p=[0.65,0.35])
distance_km   = np.random.exponential(12, n).clip(1, 80).round(1)

# Attrition is the TARGET — influenced by satisfaction, overtime, salary
attrition_prob = (
    0.05
    + (5 - satisfaction) * 0.06
    + overtime * 0.12
    + (salary < 45000) * 0.10
    + (years_company < 2) * 0.08
)
attrition = (np.random.rand(n) < attrition_prob.clip(0, 0.85)).astype(int)

df = pd.DataFrame({
    'employee_id':   range(1001, 1001 + n),
    'age':           age,
    'department':    dept_arr,
    'education':     edu_arr,
    'job_level':     level_arr,
    'years_at_company': years_company,
    'salary':        salary,
    'satisfaction':  satisfaction,
    'perf_rating':   perf_rating,
    'overtime':      overtime,
    'distance_km':   distance_km,
    'attrition':     attrition,
})

# Inject realistic messiness —————————————————————————
miss_idx = np.random.choice(n, 38, replace=False)
df.loc[miss_idx[:18], 'salary']       = np.nan
df.loc[miss_idx[18:30], 'satisfaction'] = np.nan
df.loc[miss_idx[30:], 'distance_km']  = np.nan

# A few outliers
df.loc[5,  'salary']      = 9999999
df.loc[17, 'distance_km'] = 850
df.loc[23, 'age']         = 7      # impossible

# A few duplicates
df = pd.concat([df, df.iloc[[10, 25, 40]]], ignore_index=True)

# Save
df.to_csv('employee_attrition.csv', index=False)
print(f'✅  Dataset created: {df.shape[0]} rows × {df.shape[1]} columns')
print(f'    Target: attrition (0=Stayed, 1=Left)')
print(f'    Injected: 38 missing values, 3 outliers, 3 duplicate rows')
print(f'\nFirst 5 rows:')
df.head()
````

# 🔍 Section 3: Exploratory Data Analysis (EDA)
#### *Understand before you model*

---

**Exploratory Data Analysis** is the structured process of interrogating a dataset  
before any modelling takes place. It is not optional — it is professional due diligence.

### The EDA Checklist

| Step | Question You Are Answering |
|------|---------------------------|
| **Shape & Schema** | How big is the dataset? What are the columns and types? |
| **Target Analysis** | What are we predicting? Is it balanced? |
| **Missing Values** | Where is data absent, and how much? |
| **Distributions** | What does each feature look like statistically? |
| **Outliers** | Are there values that are extreme or impossible? |
| **Correlations** | Which features relate to each other, and to the target? |
| **Categorical Profiles** | What are the unique values and their frequencies? |
| **Cross-tabulations** | How does the target vary across categories? |

Good EDA will tell you what to clean, what to engineer, and what to be suspicious of.


````python
# ── Step 1: Shape & Schema ──────────────────────────────────────────────────
df = pd.read_csv('employee_attrition.csv')

print('='*55)
print('  DATASET OVERVIEW')
print('='*55)
print(f'  Rows:     {df.shape[0]}')
print(f'  Columns:  {df.shape[1]}')
print(f'  Memory:   {df.memory_usage(deep=True).sum() / 1024:.1f} KB')

print('\n  COLUMN SCHEMA')
print(f"  {'Column':<22} {'Dtype':<12} {'Non-Null':>10} {'Null':>8} {'Null%':>8}")
print('  ' + '-'*62)
for col in df.columns:
    nn   = df[col].notna().sum()
    null = df[col].isna().sum()
    pct  = null / len(df) * 100
    flag = ' ⚠️' if pct > 0 else ''
    print(f"  {col:<22} {str(df[col].dtype):<12} {nn:>10} {null:>8} {pct:>7.1f}%{flag}")
````

````python
# ── Step 2: Target Variable Analysis ────────────────────────────────────────
print('TARGET VARIABLE: attrition')
print('-'*35)
counts    = df['attrition'].value_counts()
pcts      = df['attrition'].value_counts(normalize=True) * 100
labels    = {0: 'Stayed', 1: 'Left'}

for val in [0, 1]:
    bar = '█' * int(pcts[val] / 2)
    print(f'  {labels[val]:8s} ({val}): {bar:<40} {counts[val]:>4} ({pcts[val]:.1f}%)')

ratio = counts[0] / counts[1]
print(f'\n  Class ratio  (Stayed:Left) = {ratio:.1f}:1')
if ratio > 3:
    print('  ⚠️  Imbalanced dataset — we will discuss class imbalance handling in Meeting 5')
else:
    print('  ✅ Reasonably balanced')
````

````python
# ── Step 3: Statistical Summary by Variable Type ────────────────────────────
numeric_cols     = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include='object').columns.tolist()

print(f'Numeric columns ({len(numeric_cols)}):     {numeric_cols}')
print(f'Categorical columns ({len(categorical_cols)}): {categorical_cols}')

print('\n── Numeric Summary ─────────────────────────────────────────')
print(df[numeric_cols].describe().round(2).to_string())

print('\n── Categorical Summary ─────────────────────────────────────')
for col in categorical_cols:
    vc = df[col].value_counts()
    print(f'\n  {col} ({df[col].nunique()} unique):')
    for val, cnt in vc.items():
        bar = '▪' * int(cnt / 10)
        print(f'    {val:<20} {bar:<20} {cnt}')
````

````python
# ── Step 4: Correlation Analysis ────────────────────────────────────────────
corr_cols = ['age','years_at_company','salary','satisfaction',
             'perf_rating','overtime','distance_km','attrition']
corr = df[corr_cols].corr()

print('Correlation with Target (attrition):')
print('-'*40)
target_corr = corr['attrition'].drop('attrition').sort_values(key=abs, ascending=False)
for feature, val in target_corr.items():
    direction = '(+)' if val > 0 else '(-)'
    bar_len   = int(abs(val) * 30)
    bar       = '█' * bar_len
    print(f'  {feature:<20} {direction} {bar:<30} {val:+.3f}')

print('\n  Positive (+): feature increases → attrition risk increases')
print('  Negative (-): feature increases → attrition risk decreases')
````

# 🧹 Section 4: Data Cleaning — Missing Values
#### *The most common data quality problem you will face*

---

Missing data is universal in real-world datasets. The question is never  
*'is there missing data?'* — it's *'why is it missing, and what do we do about it?'*

### Three Types of Missingness

| Type | Meaning | Example | Implication |
|------|---------|---------|-------------|
| **MCAR** — Missing Completely At Random | Missingness has no pattern | Survey response accidentally lost | Safe to drop rows |
| **MAR** — Missing At Random | Missingness depends on other observed variables | Salary missing more often for younger employees | Impute using related features |
| **MNAR** — Missing Not At Random | Missingness relates to the value itself | High earners decline to share income | Most dangerous — may introduce bias |

### Strategies for Handling Missing Values

| Strategy | When to Use | Risk |
|----------|------------|------|
| **Drop rows** | Very few missing (<5%), MCAR | Lose data |
| **Drop columns** | >60% of column is missing | Lose feature |
| **Mean/Median imputation** | Numeric, roughly symmetric distribution | Reduces variance |
| **Mode imputation** | Categorical features | May over-represent one class |
| **Forward/Backward fill** | Time-series data | Assumes adjacent values similar |
| **Model-based imputation** | Complex relationships between features | Computationally expensive |
| **Flag + impute** | When missingness itself is informative | Best practice |

> 🔑 **Pro tip:** Always create a `_was_missing` flag column *before* imputing.  
> The fact that a value was missing can itself be a predictive signal for your model.


````python
# ── Detecting Missing Values ─────────────────────────────────────────────────
print('=== Missing Value Audit ===')
missing_df = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_pct':   (df.isnull().sum() / len(df) * 100).round(2),
    'dtype':         df.dtypes
}).query('missing_count > 0').sort_values('missing_pct', ascending=False)

print(missing_df.to_string())
print(f'\nTotal missing cells: {df.isnull().sum().sum()}')
print(f'Total cells:         {df.size}')
print(f'Overall missing:     {df.isnull().sum().sum() / df.size * 100:.2f}%')
````

````python
# ── Handling Missing Values: Step-by-step ───────────────────────────────────
df_clean = df.copy()   # Always work on a copy!

# ── 1. Flag missing before imputing ──────────────────────────────────────────
for col in ['salary', 'satisfaction', 'distance_km']:
    df_clean[f'{col}_was_missing'] = df_clean[col].isnull().astype(int)

# ── 2. Numeric: impute with median (robust to outliers) ───────────────────────
salary_median = df_clean['salary'].median()
sat_median    = df_clean['satisfaction'].median()
dist_median   = df_clean['distance_km'].median()

df_clean['salary']       = df_clean['salary'].fillna(salary_median)
df_clean['satisfaction'] = df_clean['satisfaction'].fillna(sat_median)
df_clean['distance_km']  = df_clean['distance_km'].fillna(dist_median)

print('=== After Missing Value Treatment ===')
print(f'Remaining missing values: {df_clean.isnull().sum().sum()}')
print(f'  salary imputed with median:       ${salary_median:,.0f}')
print(f'  satisfaction imputed with median:  {sat_median}')
print(f'  distance_km imputed with median:   {dist_median} km')
print(f'\nNew flag columns added:')
flag_cols = [c for c in df_clean.columns if 'was_missing' in c]
print(f'  {flag_cols}')
print(f'  Salary was missing for {df_clean["salary_was_missing"].sum()} employees')
````

````python
# ── Advanced: Group-wise imputation ──────────────────────────────────────────
# Instead of global median, impute salary by department median
# (more realistic — a Sales employee's missing salary shouldn't be
#  filled with an Engineering median)

df_clean2 = df.copy()
df_clean2['salary_was_missing'] = df_clean2['salary'].isnull().astype(int)

# Group-wise median imputation
df_clean2['salary'] = df_clean2.groupby('department')['salary'].transform(
    lambda x: x.fillna(x.median())
)

# Show department medians used for imputation
print('Department salary medians used for imputation:')
dept_medians = df.groupby('department')['salary'].median().sort_values(ascending=False)
for dept, med in dept_medians.items():
    print(f'  {dept:<15} ${med:>10,.0f}')

print(f'\nGlobal median would have been: ${df["salary"].median():,.0f}')
print('Group-wise imputation is more accurate when feature distributions differ by group.')
````

# ⚠️ Section 5: Outliers & Data Inconsistencies
#### *Values that break your model silently*

---

An **outlier** is a data point that lies far from the typical range of a feature.  
Outliers can be:

- 🐛 **Data entry errors** — salary entered as `9999999` instead of `99999`
- 🚫 **Domain violations** — age of `7` for an employee, distance of `850 km` to office
- ✅ **Genuine extreme values** — a CEO's salary being legitimately much higher than peers
- 🔍 **Interesting signals** — fraud, medical anomalies, machine failures

**Never remove outliers blindly.** First understand *why* they exist.

### Detection Methods

| Method | How It Works | Best For |
|--------|-------------|----------|
| **IQR Rule** | Outlier if below Q1−1.5×IQR or above Q3+1.5×IQR | General-purpose, robust |
| **Z-Score** | Outlier if more than 3 standard deviations from mean | Normally distributed data |
| **Domain Rules** | Define valid ranges from subject-matter knowledge | Always apply first |
| **Visual inspection** | Boxplots, scatter plots, histograms | Always do this too |


````python
# ── Outlier Detection: IQR Method ───────────────────────────────────────────
def detect_outliers_iqr(series, name=''):
    """Return outlier stats using the IQR method."""
    Q1  = series.quantile(0.25)
    Q3  = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    mask  = (series < lower) | (series > upper)
    print(f'  {name:<20} Q1={Q1:>10.1f}  Q3={Q3:>10.1f}  '
          f'Lower={lower:>10.1f}  Upper={upper:>10.1f}  '
          f'Outliers={mask.sum():>3}')
    return mask, lower, upper

print('=== IQR Outlier Detection ===')
print(f"  {'Feature':<20} {'Q1':>12} {'Q3':>12} {'Lower':>12} {'Upper':>12} {'Outliers':>10}")
print('  ' + '-'*80)
for col in ['age', 'salary', 'distance_km', 'years_at_company']:
    detect_outliers_iqr(df[col].dropna(), col)

# Domain knowledge rules
print('\n=== Domain Rules ===')
invalid_age  = df[df['age'] < 18]
invalid_sal  = df[df['salary'] > 500000]
invalid_dist = df[df['distance_km'] > 200]
print(f'  Employees with age < 18:      {len(invalid_age)}  (impossible — check: {invalid_age[["employee_id","age"]].values})')
print(f'  Salary > $500,000:            {len(invalid_sal)}  (entry error likely)')
print(f'  Distance > 200 km:            {len(invalid_dist)}  (likely error)')
````

````python
# ── Outlier Treatment ────────────────────────────────────────────────────────
df_clean = df_clean.copy()

# Strategy 1: Remove impossible values (domain rules)
before = len(df_clean)
df_clean = df_clean[df_clean['age'] >= 18]          # Remove impossible ages
print(f'Removed {before - len(df_clean)} rows with impossible age values')

# Strategy 2: Cap (Winsorise) at IQR boundaries — preserves row, reduces influence
for col in ['salary', 'distance_km']:
    Q1    = df_clean[col].quantile(0.25)
    Q3    = df_clean[col].quantile(0.75)
    IQR   = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    n_capped = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    df_clean[col] = df_clean[col].clip(lower, upper)
    print(f'Capped {n_capped} outliers in {col!r} to [{lower:.0f}, {upper:.0f}]')

# Strategy 3: Remove exact duplicates
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f'Removed {before - len(df_clean)} duplicate rows')

print(f'\nFinal clean dataset: {df_clean.shape[0]} rows × {df_clean.shape[1]} columns')
````

# ⚙️ Section 6: Feature Engineering
#### *Turning raw data into model-ready signals*

---

**Feature engineering** is the craft of creating, transforming, or selecting input variables  
that give your model the best possible information to learn from.

It is part science, part art, and heavily domain-dependent.  
A great feature engineer with a simple model will often outperform  
a poor feature engineer with a complex model.

### Core Techniques

| Technique | What It Does | Example |
|-----------|-------------|--------|
| **Binning** | Convert continuous → ordinal categories | Age → [Young, Mid, Senior] |
| **Scaling** | Normalise numeric range | Income 25k–180k → 0.0–1.0 |
| **Ratio features** | Combine two columns | Salary / Age → earnings rate |
| **Interaction features** | Multiply or combine signals | Overtime × LowSatisfaction |
| **Encoding categoricals** | Make strings numeric | Dept → dummy variables |
| **Date decomposition** | Extract sub-components from dates | Date → Year, Month, DayOfWeek |
| **Log transform** | Compress skewed distributions | log(salary) |
| **Polynomial features** | Capture non-linear relationships | Age², Age×Salary |

> 🔑 **Rule of thumb:** Features that a domain expert would consider useful  
> are usually more predictive than blindly generated combinations.


````python
# ── Feature Engineering on the Employee Dataset ──────────────────────────────
df_fe = df_clean.copy()

# ── 1. Binning ────────────────────────────────────────────────────────────────
df_fe['age_group'] = pd.cut(
    df_fe['age'],
    bins  = [17, 25, 35, 45, 55, 100],
    labels= ['22-25', '26-35', '36-45', '46-55', '56+']
)

df_fe['tenure_group'] = pd.cut(
    df_fe['years_at_company'],
    bins  = [-1, 1, 3, 7, 100],
    labels= ['New(<1yr)', 'Early(1-3)', 'Established(3-7)', 'Veteran(7+)']
)

# ── 2. Ratio features ─────────────────────────────────────────────────────────
df_fe['salary_per_year'] = (df_fe['salary'] / df_fe['years_at_company'].replace(0, 0.5)).round(0)
df_fe['salary_age_ratio'] = (df_fe['salary'] / df_fe['age']).round(0)

# ── 3. Interaction features ───────────────────────────────────────────────────
# Disengaged overworker — high-risk attrition profile
df_fe['disengaged_overworker'] = (
    (df_fe['overtime'] == 1) & (df_fe['satisfaction'] <= 2)
).astype(int)

# ── 4. Log transformation (right-skewed feature) ──────────────────────────────
df_fe['log_salary'] = np.log1p(df_fe['salary'])   # log1p = log(x+1) avoids log(0)

# ── 5. Encoding categoricals ──────────────────────────────────────────────────
# Ordinal encoding for education level
edu_order = {'High School': 1, 'Bachelor': 2, 'Master': 3, 'PhD': 4}
df_fe['education_encoded'] = df_fe['education'].map(edu_order)

# One-hot encoding for department (nominal)
dept_dummies = pd.get_dummies(df_fe['department'], prefix='dept', drop_first=True)
df_fe = pd.concat([df_fe, dept_dummies], axis=1)

print('New features created:')
new_feats = ['age_group','tenure_group','salary_per_year','salary_age_ratio',
             'disengaged_overworker','log_salary','education_encoded']
print(df_fe[new_feats + ['attrition']].head(8).to_string())
print(f'\nDataFrame shape: {df_fe.shape}')
````

````python
# ── Scaling: Min-Max vs Standard Scaler ─────────────────────────────────────
# Scaling ensures no single feature dominates due to its numeric range.
# Critical for distance-based models (KNN, SVM) and gradient-based (Neural Networks).
# Tree-based models (Decision Tree, Random Forest) do NOT need scaling.

from sklearn.preprocessing import MinMaxScaler, StandardScaler

scale_cols = ['age', 'salary', 'distance_km', 'years_at_company']
sample     = df_fe[scale_cols].dropna().head(5)

mm_scaled  = MinMaxScaler().fit_transform(sample)
std_scaled = StandardScaler().fit_transform(sample)

print('Original values (first 5 rows):')
print(sample.to_string(), '\n')

print('Min-Max scaled (range 0-1):')
print(pd.DataFrame(mm_scaled, columns=scale_cols).round(4).to_string(), '\n')

print('Standard scaled (mean=0, std=1):')
print(pd.DataFrame(std_scaled, columns=scale_cols).round(4).to_string())

print('\nWhen to use which:')
print('  MinMaxScaler  → Neural networks, when you need values in [0,1]')
print('  StandardScaler→ Most ML models — especially SVM, Logistic Regression, KNN')
print('  Neither       → Tree-based models (Decision Tree, Random Forest, XGBoost)')
````

# 📊 Section 7: Data Visualisation
#### *Making the invisible visible*

---

Numbers tell you *what*. Visualisations tell you *why* and *how*.  
In EDA, a good chart will reveal patterns that summary statistics completely hide.

### The Visualisation Toolkit

| Chart Type | Use Case | Seaborn / Matplotlib |
|-----------|----------|---------------------|
| **Histogram** | Distribution of a single numeric variable | `sns.histplot` |
| **Box Plot** | Distribution + outliers, compare groups | `sns.boxplot` |
| **Bar Chart** | Category counts or means | `sns.barplot / countplot` |
| **Scatter Plot** | Relationship between two numeric features | `sns.scatterplot` |
| **Heatmap** | Correlation matrix | `sns.heatmap` |
| **Violin Plot** | Distribution shape across groups | `sns.violinplot` |
| **Pair Plot** | All pairwise relationships at once | `sns.pairplot` |
| **Line Plot** | Trends over time | `plt.plot / sns.lineplot` |

> 🔑 **Always label your axes and title your plots.**  
> A chart without labels is useless in a professional context.


````python
# ── Plot 1: Distribution of Salary (raw vs log-transformed) ─────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Raw salary
sns.histplot(df_fe['salary'].dropna(), bins=40, kde=True,
             color='#0D7377', ax=axes[0])
axes[0].set_title('Salary Distribution (Raw)', fontweight='bold')
axes[0].set_xlabel('Salary ($)')
axes[0].set_ylabel('Count')
axes[0].xaxis.set_major_formatter(mtick.FuncFormatter(lambda x,_: f'${x/1000:.0f}k'))

# Log-transformed salary
sns.histplot(df_fe['log_salary'].dropna(), bins=40, kde=True,
             color='#F0A500', ax=axes[1])
axes[1].set_title('Log(Salary) Distribution — More Symmetric', fontweight='bold')
axes[1].set_xlabel('log(Salary)')
axes[1].set_ylabel('Count')

plt.suptitle('Effect of Log Transformation on Skewed Features',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print('Left: right-skewed (pulled by high earners)')
print('Right: log-transformed — more bell-shaped, better for many models')
````

````python
# ── Plot 2: Attrition vs Key Numeric Features ────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

plot_features = ['salary', 'satisfaction', 'age', 'distance_km']
palette = {0: '#0D7377', 1: '#F0A500'}

for i, feat in enumerate(plot_features):
    sns.boxplot(
        data    = df_fe.dropna(subset=[feat]),
        x       = 'attrition',
        y       = feat,
        palette = palette,
        ax      = axes[i],
        width   = 0.5
    )
    axes[i].set_title(f'{feat.replace("_"," ").title()} by Attrition Status',
                      fontweight='bold')
    axes[i].set_xlabel('')
    axes[i].set_xticklabels(['Stayed (0)', 'Left (1)'])
    axes[i].set_ylabel(feat.replace('_', ' ').title())

plt.suptitle('Feature Distributions: Stayed vs Left',
             fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print('Observations:')
for feat in plot_features:
    stayed = df_fe[df_fe['attrition']==0][feat].median()
    left   = df_fe[df_fe['attrition']==1][feat].median()
    diff   = ((left - stayed) / stayed * 100)
    print(f'  {feat:<16} Stayed={stayed:>8.1f}  Left={left:>8.1f}  Diff={diff:+.1f}%')
````

````python
# ── Plot 3: Categorical Features vs Attrition ────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

cat_feats = ['department', 'job_level', 'satisfaction']

for i, feat in enumerate(cat_feats):
    attr_rate = df_fe.groupby(feat)['attrition'].mean().sort_values(ascending=True)
    colors    = ['#F0A500' if v > 0.25 else '#0D7377' for v in attr_rate.values]

    bars = axes[i].barh(attr_rate.index.astype(str), attr_rate.values,
                        color=colors, edgecolor='white', height=0.6)
    axes[i].axvline(df_fe['attrition'].mean(), color='red',
                    linestyle='--', linewidth=1.5, label='Overall avg')
    axes[i].set_xlabel('Attrition Rate')
    axes[i].set_title(f'Attrition Rate by\n{feat.replace("_"," ").title()}',
                      fontweight='bold')
    axes[i].xaxis.set_major_formatter(mtick.PercentFormatter(1.0))
    axes[i].legend(fontsize=8)

    for bar, val in zip(bars, attr_rate.values):
        axes[i].text(val + 0.005, bar.get_y() + bar.get_height()/2,
                     f'{val:.0%}', va='center', fontsize=9)

plt.suptitle('Which Groups Have the Highest Attrition Risk?',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# ── Plot 4: Correlation Heatmap ───────────────────────────────────────────────
corr_feats = ['age', 'years_at_company', 'salary', 'satisfaction',
              'perf_rating', 'overtime', 'distance_km',
              'disengaged_overworker', 'log_salary', 'attrition']

corr_matrix = df_fe[corr_feats].corr()

fig, ax = plt.subplots(figsize=(12, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # hide upper triangle

sns.heatmap(
    corr_matrix, mask=mask, annot=True, fmt='.2f',
    cmap='RdYlGn', center=0, vmin=-1, vmax=1,
    linewidths=0.5, linecolor='white',
    annot_kws={'size': 9}, ax=ax
)
ax.set_title('Feature Correlation Matrix\n(Green = positive, Red = negative)',
             fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()

print('Reading the heatmap:')
print('  • Each cell shows correlation between two features (-1 to +1)')
print('  • Bottom row: correlation of each feature WITH the target (attrition)')
print('  • Features highly correlated with EACH OTHER may cause multicollinearity')
print('  • Look for the darkest cells in the attrition row — those are your best features')
````

````python
# ── Plot 5: Salary Distribution by Department and Attrition ──────────────────
fig, ax = plt.subplots(figsize=(14, 6))

sns.violinplot(
    data    = df_fe.dropna(subset=['salary']),
    x       = 'department',
    y       = 'salary',
    hue     = 'attrition',
    split   = True,
    palette = {0: '#0D7377', 1: '#F0A500'},
    inner   = 'quartile',
    ax      = ax
)
ax.set_title('Salary Distribution by Department (Split by Attrition)',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Department')
ax.set_ylabel('Salary ($)')
ax.yaxis.set_major_formatter(mtick.FuncFormatter(lambda x,_: f'${x/1000:.0f}k'))
ax.legend(title='Attrition', labels=['Stayed','Left'])
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

print('Violin plot reads:')
print('  • Width = density (how many people at that salary level)')
print('  • Left half = Stayed, Right half = Left (split violin)')
print('  • Inner lines = quartiles (Q1, Median, Q3)')
````

# 🔗 Section 8: The Complete EDA Pipeline
#### *Everything in one reusable workflow*

---

This section assembles everything from today into a single, reusable EDA pipeline function.  
In practice, you will adapt this pattern for every new dataset you encounter.

**The Professional EDA Workflow:**
```
Load → Schema → Target → Quality → Distributions → Correlations → Visualise → Report
```

The function below can be dropped into any notebook and run on any tabular dataset.


````python
def run_eda(filepath, target_col, numeric_plot_cols=None, cat_plot_cols=None):
    """
    Run a comprehensive Exploratory Data Analysis on a CSV dataset.

    Args:
        filepath (str):         Path to CSV file
        target_col (str):       Name of the target/label column
        numeric_plot_cols (list): Numeric columns to visualise (optional)
        cat_plot_cols (list):    Categorical columns to visualise (optional)

    Returns:
        pd.DataFrame: Cleaned and feature-engineered DataFrame
    """
    df = pd.read_csv(filepath)

    # ── 1. Overview ───────────────────────────────────────────────────────────
    print('='*60)
    print('  AUTOMATED EDA REPORT')
    print('='*60)
    print(f'  File:     {filepath}')
    print(f'  Target:   {target_col}')
    print(f'  Shape:    {df.shape[0]:,} rows × {df.shape[1]} columns')
    print(f'  Memory:   {df.memory_usage(deep=True).sum()/1024:.1f} KB')

    # ── 2. Data Types ─────────────────────────────────────────────────────────
    num_cols = df.select_dtypes(include=np.number).columns.tolist()
    cat_cols = df.select_dtypes(include='object').columns.tolist()
    print(f'\n  Numeric features ({len(num_cols)}):     {num_cols}')
    print(f'  Categorical features ({len(cat_cols)}): {cat_cols}')

    # ── 3. Target Distribution ────────────────────────────────────────────────
    print(f'\n  TARGET: {target_col}')
    vc   = df[target_col].value_counts()
    vpct = df[target_col].value_counts(normalize=True) * 100
    for val in vc.index:
        print(f'    Class {val}: {vc[val]:>5} ({vpct[val]:.1f}%)')

    # ── 4. Missing Values ─────────────────────────────────────────────────────
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    print(f'\n  MISSING VALUES ({missing.sum()} total):')
    if len(missing) == 0:
        print('    None detected ✅')
    else:
        for col, cnt in missing.items():
            print(f'    {col:<22}: {cnt:>4} ({cnt/len(df)*100:.1f}%)')

    # ── 5. Duplicates ─────────────────────────────────────────────────────────
    dupes = df.duplicated().sum()
    print(f'\n  DUPLICATES: {dupes} rows')

    # ── 6. Target Correlation ─────────────────────────────────────────────────
    print(f'\n  FEATURE CORRELATION WITH TARGET ({target_col}):')
    corr = df[num_cols].corr()[target_col].drop(target_col).sort_values(key=abs, ascending=False)
    for feat, val in corr.head(6).items():
        bar = '█' * int(abs(val) * 25)
        print(f'    {feat:<22} {bar:<25} {val:+.3f}')

    # ── 7. Visualisations ────────────────────────────────────────────────────
    _num  = numeric_plot_cols or num_cols[:4]
    _cat  = cat_plot_cols     or cat_cols[:2]

    fig, axes = plt.subplots(2, 2, figsize=(14, 9))
    axes = axes.flatten()

    for i, feat in enumerate(_num[:4]):
        if feat == target_col: continue
        df.boxplot(column=feat, by=target_col, ax=axes[i], grid=False)
        axes[i].set_title(f'{feat} by {target_col}', fontweight='bold')
        axes[i].set_xlabel(target_col)
    plt.suptitle('EDA: Feature Distributions by Target Class',
                 fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.show()

    print('\n' + '='*60)
    print('  EDA COMPLETE')
    print('='*60)
    return df


# ── Run it on our dataset ─────────────────────────────────────────────────────
result = run_eda(
    filepath       = 'employee_attrition.csv',
    target_col     = 'attrition',
    numeric_plot_cols = ['salary', 'satisfaction', 'age', 'distance_km'],
    cat_plot_cols     = ['department', 'job_level']
)
````
